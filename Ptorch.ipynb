Gradient Computation for Linear Regression


```python
import torch
```

```python
def compute_mse_gradients(x,y_true):
#initialize weight and bias with requires_grad = True
  w = torch.ones(1, requires_grad=True)
  b = torch.zeros(1, requires_grad=True)
#Forward pass: compute predicted y
  y_pred = w * x + b
  loss = torch.mean(y_pred - y_true)**2
  loss.backward()
  return w.grad, b.grad
```

```python
x = torch.tensor([1.0,2.0,3.0])
y_true = torch.tensor([2.0,4.0,6.0])

w_grad, b_grad = compute_mse_gradients(x,y_true)
print(f"Gradient of w:{w_grad}")
print(f"Gradient of b:{b_grad}")
```
