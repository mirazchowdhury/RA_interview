# Hessian matrix

```python
import torch
```

```python
def approximate_hessian(A,b,c,w, epsilon=1e-4):
  #define the quadraqtic function
  def f(w_vec):
    return torch.matmul(torch.matmul(w_vec.T, A),w_vec)+torch.matmul(b.T,w_vec)+c

    #compute gradient at w
    w_tensor = w.clone().requires_grad_(True)
    loss = f(w_tensor)
    loss.backward()
    grad_w = w_tensor.grad.clone()

    #Initialize Hessian matrix
    n = len(w)
    hessian = torch.zeros((n,n))

    #Compute Hessian using finite differences
    for i in range(n):
      # Create perturbation vector
      e_i = torch.zeros_like(w)
      e_i[i] = epsilon
      #Compute gradient at w+epsilon*e_i
      w_perturbed = w.clone() + e_i
      w_perturbed.requires_grad_(True)
      loss_perturbed = f(w_perturbed)
      loss_perturbed.backward()
      grad_w_perturbed = w_perturbed.grad.clone()

      #Estimate Hessian column using finite differences
      hessian[:,i] = (grad_w_perturbed - grad_w)/epsilon

    return hessian
```

```python
A = torch.tensor([[2.0,1.0],[1.0,2.0]])
b = torch.tensor([1.0,1.0])
c = 0.0
w = torch.tensor([1.0,1.0])
hessian = approximate_hessian(A,b,c,w)
print(hessian)
print(2*A) #True for quadrate function
```
